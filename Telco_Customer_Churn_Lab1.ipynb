```python
# Lab 1: Telco Customer Churn Analysis
# Author: Pacifique Uwineza (Reg No: 224005544)
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
# Load the dataset
df = pd.read_csv('WA_Fn-UseC_-Telco-Customer-Churn.csv')
print("Dataset loaded successfully with shape:", df.shape)
```

```python
# Display the first 5 rows
print("--- Dataset Head ---")
display(df.head())

# Display the last 5 rows
print("--- Dataset Tail ---")
display(df.tail())
```

```python
# 1. Change Churn yes/no to 1/0
df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})

# 2. Check datatypes and fix TotalCharges (which has hidden spaces)
print("--- Datatypes Before Correction ---")
print(df.dtypes)

df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

# 3. Display null values and remove them
print("\n--- Null Values Before Removal ---")
print(df.isnull().sum())

df = df.dropna()
print("\nDataset shape after removing nulls:", df.shape)
```

```python
print("--- Gender vs Churn (%) ---")
print(pd.crosstab(df['gender'], df['Churn'], normalize='index') * 100)

print("\n--- SeniorCitizen vs Churn (%) ---")
print(pd.crosstab(df['SeniorCitizen'], df['Churn'], normalize='index') * 100)

print("\n--- Partner vs Churn (%) ---")
print(pd.crosstab(df['Partner'], df['Churn'], normalize='index') * 100)

print("\n--- Dependents vs Churn (%) ---")
print(pd.crosstab(df['Dependents'], df['Churn'], normalize='index') * 100)
```

```python
print("--- PhoneService vs Churn (%) ---")
print(pd.crosstab(df['PhoneService'], df['Churn'], normalize='index') * 100)
```
